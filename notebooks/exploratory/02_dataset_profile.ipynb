# Dataset profiling
What the compiler sees before it trains anything: intrinsic dimension, autocorrelation, noise, chaos hints.

```python
from nssc.data.builder import build_dataset
from nssc.utils.config import load_yaml
from nssc.compiler.profiler import profile_dataset
ds = build_dataset(load_yaml('../../configs/datasets/lorenz63.yaml'))
prof = profile_dataset(ds)
print(prof.to_markdown())
```

```python
import matplotlib.pyplot as plt
plt.plot(prof.autocorr); plt.xlabel('lag'); plt.ylabel('mean ACF'); plt.title(f'autocorr time = {prof.autocorr_time}')
```
