# Synthetic dynamical systems
Quick look at every registered generator (Tier 1) and a Tier-2 observation map.

```python
import numpy as np, matplotlib.pyplot as plt
import nssc.data
from nssc.utils.registry import SYSTEMS
print(SYSTEMS.keys())
```

```python
fig, axes = plt.subplots(3, 4, figsize=(14, 9))
for ax, k in zip(axes.ravel(), SYSTEMS.keys()):
    x = SYSTEMS.build(k).simulate(1, 800, seed=0)[0]
    ax.plot(x[:, :4]); ax.set_title(k)
plt.tight_layout()
```

```python
from nssc.data.builder import build_dataset
ds = build_dataset({'system': 'lorenz63', 'n_traj': 4, 'n_steps': 500, 'dt': 0.01, 'transient': 1000,
                    'observation': {'type': 'mlp', 'obs_dim': 32, 'seed': 0}, 'noise_std': 0.05})
print(ds.x.shape, ds.z_true.shape)
plt.plot(ds.x[0, :, :6]); plt.title('Lorenz-63 through a random MLP (6 of 32 dims)')
```
