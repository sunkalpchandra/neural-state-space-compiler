# Compiler walkthrough (tiny config)
Runs in ~10 s on CPU. Inspect candidates, ranking terms and the report.

```python
import torch
from nssc.utils.config import load_config
from nssc.compiler import StateSpaceCompiler
from nssc.utils.experiment_registry import ExperimentRegistry
cfg = load_config('../../configs/compiler/tiny.yaml'); cfg['output_dir'] = '/tmp/nssc_nb_compile'
comp = StateSpaceCompiler(cfg, device=torch.device('cpu'), registry=ExperimentRegistry('/tmp/nssc_nb_registry.jsonl'))
cm = comp.run()
print(cm.report.to_markdown())
```

```python
import pandas as pd
rows = [{'candidate': r['name'], 'J': r['score'], **r['terms']} for r in cm.report.ranking]
pd.DataFrame(rows)
```

```python
from nssc.experiment import prepare_data
splits, stats, raw = prepare_data(dict(comp.cfg['dataset']))
x = torch.from_numpy(splits['test'].x[:1])
x_hat, z_hat = cm.rollout(x[:, :8], 40)
import matplotlib.pyplot as plt
plt.plot(x[0, 8:48, 0], label='true'); plt.plot(x_hat[0, :, 0].detach(), '--', label='rollout'); plt.legend()
```
